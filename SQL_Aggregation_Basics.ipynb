```python

import sqlite3
import pandas as pd

# Create an in-memory SQLite database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create a Sales table
cursor.execute("""
CREATE TABLE Sales (
    SaleID INTEGER,
    Product TEXT,
    Category TEXT,
    Amount REAL,
    Quantity INTEGER
)
""")

sales_data = [
    (1, 'Laptop', 'Electronics', 50000, 2),
    (2, 'Mouse', 'Electronics', 800, 5),
    (3, 'Keyboard', 'Electronics', 1500, 3),
    (4, 'Chair', 'Furniture', 4500, 4),
    (5, 'Desk', 'Furniture', 9000, 2),
    (6, 'Notebook', 'Stationery', 100, 10),
    (7, 'Pen', 'Stationery', 20, 20),
    (8, 'Monitor', 'Electronics', 12000, 2),
    (9, 'Table', 'Furniture', None, 1),
    (10, 'Lamp', 'Furniture', 1200, 3)
]

cursor.executemany(
    "INSERT INTO Sales VALUES (?, ?, ?, ?, ?)",
    sales_data
)
conn.commit()

print("Sales table created successfully!")
display(pd.read_sql_query("SELECT * FROM Sales", conn))
```

```python
query = """
SELECT COUNT(*) AS TotalRecords
FROM Sales;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT SUM(Amount) AS TotalSales
FROM Sales;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT AVG(Amount) AS AverageSales
FROM Sales;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT MIN(Amount) AS MinimumSale
FROM Sales;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT MAX(Amount) AS MaximumSale
FROM Sales;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Category, COUNT(*) AS TotalRecords
FROM Sales
GROUP BY Category;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Category, SUM(Amount) AS CategorySales
FROM Sales
GROUP BY Category;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Category, AVG(Amount) AS AverageSale
FROM Sales
GROUP BY Category;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    Category,
    MIN(Amount) AS MinimumSale,
    MAX(Amount) AS MaximumSale
FROM Sales
GROUP BY Category;
"""
display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    COUNT(*) AS TotalRecords,
    COUNT(Amount) AS RecordsWithAmount,
    COUNT(*) - COUNT(Amount) AS MissingAmounts
FROM Sales;
"""
display(pd.read_sql_query(query, conn))
```
